```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
df = pd.read_csv("data.csv")
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df['Calories'] = df['Calories'].fillna(df['Calories'].median())
```

```python
df.isnull().sum()
```

```python
df["Date"] = df["Date"].replace("","nan")


```

```python
df = df.dropna(subset=["Date"])
```

```python
df.isnull().sum()
```

```python
filtered_df = df[df["Calories"] > 300]
filtered_df.head()
```

```python
df["Pluse_Difference"] = df["Maxpulse"] - df["Pulse"]
```

```python
df.head()
```

```python

plt.bar(df["Duration"],df["Calories"])
plt.xlabel("Duration")
plt.ylabel("Calories")
plt.title("Duration vs Calories")
plt.show()
plt.close()
```

```python
plt.scatter(df["Pulse"],df["Calories"])
plt.xlabel("Pulse")
plt.ylabel("Calories")
plt.title("Pulse vs Categories")
plt.show()
plt.close()
```

```python
sns.boxenplot(x = df["Calories"])
plt.title("Calories Distribution")
plt.show()
```
